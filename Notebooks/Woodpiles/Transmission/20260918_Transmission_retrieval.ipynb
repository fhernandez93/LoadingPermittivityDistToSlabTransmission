```python
from dotenv import load_dotenv
load_dotenv()
import tidy3d as td
from tidy3d import web
import numpy as np
from pathlib import Path
import re
```

```python
import sys, os

# AutomationModule lives in the root of the tidy3d project
sys.path.append(os.path.abspath(r'../../../../tidy3d'))
import AutomationModule as AM
tidy3dAPI = os.environ["API_TIDY3D_KEY"]
```

```python
# folder_path = rf"{Path.cwd().parents[2]}/data/20260918_Transmission_Experiment_woodpiles"
folder_path = rf"{Path.cwd().parents[2]}/data/20261017_Frank_rho_sweep_Transmission_Experiment_woodpiles"
h5_file = rf"../data/20261017_rho_sweep_n_3p6_transmission_data_woodpiles.h5"
```

```python
transmission_data = AM.read_hdf5_as_dict(h5_file)
```

```python
# Loop through all files in the folder
for dirpath, dirnames, filenames in os.walk(folder_path):
      for filename in filenames:
        #Retrieve simulation data 
        if os.path.isfile(os.path.join(dirpath, filename)):
          file=os.path.join(dirpath, filename)
          if filename in transmission_data.keys():
             continue
          structure_1 = AM.loadFromFile(key = tidy3dAPI, file_path=file,get_ref=True)
          sim_data_i,sim_data_ref = structure_1.sim_data,structure_1.sim_data0
          transmission_entry = sim_data_i['flux2'].flux/sim_data_ref['flux2'].flux
          transmission_exit = sim_data_i['flux1'].flux/sim_data_ref['flux1'].flux
          transmission_data[filename] = {'transmission_entry': transmission_entry, 'transmission_exit': transmission_exit,"lambda":td.C_0/np.array(sim_data_i.simulation.monitors[0].freqs)}
```
