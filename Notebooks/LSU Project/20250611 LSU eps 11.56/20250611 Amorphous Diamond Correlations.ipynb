```python
#Getting a values 
folder_path = r"H:\phd stuff\tidy3d\structures\Crystal\20250617_eps_9_L_20.25"
a_values = {}
for f,filename in enumerate(os.listdir(folder_path)):
    print(filename)
    if not (Path(filename).suffix==".h5" or Path(filename).suffix==".stl"):
        continue 
    if os.path.isfile(os.path.join(folder_path, filename)):
        file=os.path.join(folder_path, filename)
        ff= re.search(r'ff_0.\d+', filename).group(0)
        key=float(f"0.{ff[5:]}")
        if key not in a_values:
            a_values[key] = {}
        val = get_a_from_h5_eps(file,20.25)
        a_values[key]["a"]=val[0]
        a_values[key]["r_vals"]=val[1]
        a_values[key]["C_r"]=val[2]
                
```

```python
ff_vals_sorted=np.array(sorted(set(x for x in a_values.keys())))
```

```python
font = {
        'weight' : 'bold',
        'size'   : 15}
fig = plt.figure(figsize=(10, 5))
colors = plt.cm.gist_rainbow(np.linspace(0, 1, len(ff_vals_sorted)))

for i,item in enumerate(ff_vals_sorted):
    plt.plot(a_values[item]["r_vals"]*0.8,a_values[item]["C_r"],label=fr"ff={item}; $a^0_{{eff}}={a_values[item]["a"]*0.8:.4f} \mu m$",color=colors[i])

plt.xlabel(r'$r[\mu m]$')
plt.ylabel(r'$C_r(\delta r)$')
plt.legend(fontsize=10)
plt.title(rf"$C_r(\delta r)$ Amorphous Diamond; $d^0=0.8 \mu m$ L={20.25*0.8:.2f} um")
plt.show()
```
