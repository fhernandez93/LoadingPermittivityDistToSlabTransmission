```python
n=3.4
medium = td.Medium(permittivity=n**2)
run = True
```

```python
lambdas = np.array([7.5, 3.5])
```

```python
folder_path=rf"./Data/Akash_data"
project_name = fr"20260911_Transmission_Spheres_Pine"
runtime_ps = 20e-12
min_steps_per_lambda = 15
id0 = ""
add_ref = True
```

```python
data = [] #arrays of [x,y,z,r]
names = []
for l,file in enumerate((os.listdir(folder_path))):
    if Path(file).suffix == ".csv":
        print(Path(file).stem)
        df = pd.read_csv(os.path.join(folder_path, file))
        print(f'  Box Dimensions:')
        print(df["box_Lx"].max(),df["box_Ly"].max(),df["box_Lz"].max())
        print(f'  Sphere Number:')
        print(len(df))
        data.append(df)
        names.append(Path(file).stem)
```

```python
cuts=[0.3,0.5,0.7,1.0]
for k,item in enumerate(data):
    for cut in cuts:
        t_slabx, t_slaby, t_slabz = item["box_Lx"].max(), item["box_Ly"].max(), item["box_Lz"].max()*cut
        Lx, Ly, Lz = t_slabx, t_slaby, item["box_Lz"].max()+12
        sphere_group = []
        for x, y, z, r in zip(item['x'], item['y'], item['z'], item['radius']):
            if np.abs(z) + r >= t_slabz/2:
                continue
            for dx in (0, -t_slabx, t_slabx):
                for dy in (0, -t_slaby, t_slaby):
                    xi, yi = x + dx, y + dy
                    if abs(xi) - r < t_slabx/2 and abs(yi) - r < t_slaby/2:   # image touches the domain
                        sphere_group.append(td.Sphere(center=[xi, yi, z], radius=r))
        structure = td.Structure(geometry=td.GeometryGroup(geometries=sphere_group),  medium=td.Medium(permittivity=1))

        f_min, f_max = td.C_0/lambdas[0], td.C_0/lambdas[1]
        f0 = (f_min + f_max) / 2
        source_def = td.PlaneWave(
            source_time = td.GaussianPulse(
                freq0=f0,
                fwidth=0.3*(f_max - f_min),
                offset=10,
                phase=0
            ),
            size= (td.inf, td.inf, 0) 
                  ,
            center=( 0, 0, (-item["box_Lz"].max()/2)-3),
            direction='+',
            name='planewave',
            )
        freq_range = td.C_0/np.array(lambdas)
                                
        monitor_freqs = np.linspace(freq_range[0],freq_range[1],250)
        monitorFlux_exit = td.FluxMonitor(
                                        center=[0,0,item["box_Lz"].max()/2+2],
                                        size=[
                                                td.inf,
                                                td.inf,
                                                0
                                            ],
                                        freqs=monitor_freqs,
                                        name="exit",
                                    )
        monitorFlux_entry = td.FluxMonitor(
                                            center=[0,0,-item["box_Lz"].max()/2-2],
                                            size=[
                                                    td.inf,
                                                    td.inf,
                                                    0
                                                ],
                                            freqs=monitor_freqs,
                                            name="entry",
                                            )
        dl = (lambdas[1] /min_steps_per_lambda) / n
        boundaries= td.BoundarySpec(
                                x= td.Boundary.periodic(),
                                y= td.Boundary.periodic(),
                                z=td.Boundary(plus=td.Absorber(num_layers=200),minus=td.Absorber(num_layers=200))
                            )
        sim = td.Simulation(
                            center = (0, 0, 0),
                            size = (Lx, Ly, Lz),
                            grid_spec = td.GridSpec.auto(min_steps_per_wvl=min_steps_per_lambda,wavelength=lambdas[0],
                                        dl_min=dl,
                                        max_scale=1.2),
                            sources = [source_def],
                            monitors = [monitorFlux_exit, monitorFlux_entry],
                            run_time = runtime_ps,
                            shutoff =1e-20, #Simulation stops when field has decayed to this 
                            boundary_spec = boundaries,
                            normalize_index = None,
                            structures = [structure],
                            subpixel=True,
                            medium=medium
                            )
        fig, ax = plt.subplots(1, tight_layout=True, figsize=(16, 8))
        sim.plot(x=0, ax=ax)
        plt.show()    
        sim_name = rf"{names[k]}_size_z_{t_slabz:.4f}"
        if run:
            if add_ref:
                 id0 =web.upload(sim.copy(update={'structures':[]}), folder_name=project_name,task_name=fr"Reference", verbose=True)
                 web.start(task_id = id0)
                 web.monitor(task_id=id0,verbose=True)
                 add_ref = False
        
            id =web.upload(sim, folder_name=project_name,task_name=sim_name, verbose=True)
            web.start(task_id = id)
            web.monitor(task_id=id,verbose=True)
            ids = id0+ '\n' + id
            structure_folder =names[k]
            file_path = rf"../../data/{project_name}/{structure_folder}/{sim_name}.txt"
            # Check if the folder exists
            if not os.path.exists( fr"../../data/{project_name}/{structure_folder}"):
                os.makedirs(fr"../../data/{project_name}/{structure_folder}",exist_ok=True)
                print(fr"Folder '../../data/{project_name}/{structure_folder}' created successfully.")
        
            # Open file in write mode
            with open(file_path, "w") as f:
                # Write the string to the file
                f.write(ids)
        else:
            sim.plot_3d()
            id =web.upload(sim,task_name=sim_name, verbose=True)
            print(web.estimate_cost(id))
            web.delete(id)
            raise TypeError("Program ends here")
```
