```python
tidy3dAPI = os.environ["API_TIDY3D_KEY"]
folder_data = "./data/transmission"
os.makedirs(folder_data, exist_ok=True)
```

```python
plt.rcParams.update({'font.size': 18})  
for path_direction in [
                  rf"H:\Codes\tidy3d\data\20260911_Transmission_Spheres_Pine"
                       ]:
  data_transmission = {}
  for dirpath, dirnames, filenames in (os.walk(path_direction)):
    fig, ax = plt.subplots(figsize=(15, 7.5)) 
    for file in filenames: 
      print(file)
      Lz = re.search(r'size_z_(\d+(?:\.\d+)?)', file).group(1)
      print(f"Lz: {Lz}")
      file_i=os.path.join(dirpath,file)
      structure_name = Path(dirpath).name
      print(f"Structure Name: {structure_name}")
      structure_1 = AM.loadFromFile(key = tidy3dAPI, file_path=file_i)
      sim_data, sim_data0 = structure_1.sim_data, structure_1.sim_data0
      transmission0_exit = sim_data0['exit'].flux
      transmission_exit = sim_data['exit'].flux
      transmission0_entry = sim_data0['entry'].flux
      transmission_entry = sim_data['entry'].flux
      transmission_normalized_exit = transmission_exit / transmission0_exit
      transmission_normalized_entry = transmission_entry / transmission0_entry
      if structure_name not in data_transmission.keys():
        data_transmission[structure_name] = {}
      if Lz not in data_transmission[structure_name].keys():
        data_transmission[structure_name][Lz] = {}
      data_transmission[structure_name][Lz]["exit"] = transmission_normalized_exit.values
      data_transmission[structure_name][Lz]["entry"] = transmission_normalized_entry.values

lambdas = (td.C_0/np.array(np.array(sim_data.simulation.monitors)[0].freqs))
      
```

```python
AM.create_hdf5_from_dict({"transmission_data": data_transmission,"lambdas": lambdas}, f"{folder_data}/20260911_Spheres_Pine_Transmission.h5")
```

```python
for structure_i in data_transmission.keys():
  fig, ax = plt.subplots(figsize=(15, 7.5)) 
  for Lz_i in natsorted(data_transmission[structure_i].keys()):
    ax.plot(lambdas, data_transmission[structure_i][Lz_i]["exit"], label=f"Lz={Lz_i}um")
    # ax.plot(lambdas, data_transmission[structure_i][Lz_i]["exit"], label=f"Lz={Lz_i}")
  ax.legend(loc=1)
  ax.grid(True, which="both")
  ax.set_yscale("log")
  ax.set_xlabel(rf'$\lambda(\mu\mathrm{{m}})$')
  ax.set_ylabel("Transmission")
  ax.set_title(rf'{structure_i}')
  plt.savefig(f"{folder_data}/{structure_i}.png")
  plt.show()
```
